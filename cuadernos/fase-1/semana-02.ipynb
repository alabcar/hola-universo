## 1. Grados, minutos y segundos de arco

Un grado se divide en 60 minutos de arco (′) y cada minuto en 60 segundos de arco (″). Escribimos las dos conversiones como funciones.

```python
def a_decimal(grados, minutos=0, segundos=0):
    """De grados, minutos y segundos de arco a grados decimales."""
    return grados + minutos / 60 + segundos / 3600

def a_gms(grados_decimales):
    """De grados decimales a (grados, minutos, segundos)."""
    g = int(grados_decimales)
    resto = (grados_decimales - g) * 60
    m = int(resto)
    s = (resto - m) * 60
    return g, m, s

print(a_decimal(0, 31, 56))          # diámetro aparente del Sol, aprox.
print(a_gms(0.5178))                 # diámetro aparente medio de la Luna
```

## 2. Radianes y tamaño angular

Un **radián** es el ángulo que abarca un arco de longitud igual al radio. Una circunferencia completa son $2\pi$ radianes, es decir, 360°. Para un objeto de tamaño $d$ visto a distancia $D$, con $d \ll D$, el ángulo que abarca es aproximadamente

$$\theta \approx \frac{d}{D} \quad (\text{en radianes})$$

La fórmula exacta usa la función arcotangente, que aún no has estudiado: la incluimos como caja negra para medir cuánto se equivoca la aproximación.

```python
import numpy as np

def grados(rad):
    return np.degrees(rad)

def tamano_angular_aprox(d, D):
    """Tamaño angular en grados con la aproximación d/D (d y D en la misma unidad)."""
    return grados(d / D)

def tamano_angular_exacto(d, D):
    """Tamaño angular en grados, fórmula exacta (caja negra por ahora)."""
    return grados(2 * np.arctan(d / (2 * D)))

diametro_sol, diametro_luna, diametro_tierra = 1.3914e6, 3474, 12742   # km
ua, dist_luna = 1.496e8, 3.844e5                                        # km

for nombre, d, D in [("Sol", diametro_sol, ua),
                     ("Luna", diametro_luna, dist_luna),
                     ("Tierra vista desde la Luna", diametro_tierra, dist_luna)]:
    a, e = tamano_angular_aprox(d, D), tamano_angular_exacto(d, D)
    print(f"{nombre:28s} aprox {a:7.4f}°   exacto {e:7.4f}°   ({a*60:5.1f} ′)")
```

Las dos fórmulas coinciden hasta la cuarta cifra: para ángulos tan pequeños la aproximación es excelente. Cuando el objeto ocupa una parte grande del cielo deja de serlo. Prueba a continuación con una pared a 1 metro del ojo.

```python
print(tamano_angular_aprox(1.0, 1.0), tamano_angular_exacto(1.0, 1.0))   # objeto de 1 m a 1 m de distancia
```

## 3. ¿Eclipse total o anular?

La distancia a la Luna y al Sol varía a lo largo del tiempo, y con ellas su tamaño aparente. Con los extremos de las distancias calculamos el rango de cada uno.

```python
import matplotlib.pyplot as plt

# Distancias extremas, en km (Tierra-Sol: perihelio y afelio; Tierra-Luna: perigeo y apogeo aproximados)
sol_min, sol_max = 1.47095e8, 1.52100e8
luna_min, luna_max = 3.5641e5, 4.0674e5

rangos = {
    "Sol":  (tamano_angular_aprox(diametro_sol, sol_max)  * 60, tamano_angular_aprox(diametro_sol, sol_min)  * 60),
    "Luna": (tamano_angular_aprox(diametro_luna, luna_max) * 60, tamano_angular_aprox(diametro_luna, luna_min) * 60),
}

fig, ax = plt.subplots(figsize=(7, 2.8))
for i, (nombre, (lo, hi)) in enumerate(rangos.items()):
    ax.barh(i, hi - lo, left=lo, height=0.5)
    ax.text(hi + 0.1, i, f"{lo:.1f}′ – {hi:.1f}′", va="center")
ax.set_yticks(range(len(rangos)))
ax.set_yticklabels(rangos.keys())
ax.set_xlabel("diámetro aparente (minutos de arco)")
ax.set_xlim(28, 36)
ax.set_title("Tamaño aparente del Sol y de la Luna a lo largo del tiempo")
plt.tight_layout()
plt.show()
```

Los rangos se solapan. Cuando la Luna parece mayor que el Sol, un eclipse solar puede ser total; cuando parece menor, queda un anillo de Sol visible (eclipse anular).
