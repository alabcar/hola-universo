## 1. Velocidad media en intervalos cada vez más pequeños

Una piedra cae y su posición (distancia recorrida) es $s(t) = 4{,}9\,t^2$ metros. ¿Qué velocidad tiene exactamente en $t = 1$ s? Calculamos la velocidad media entre $t=1$ y $t = 1 + h$ para valores de $h$ cada vez más pequeños:

$$v_{\text{media}} = \frac{s(1+h) - s(1)}{h}.$$

```python
import numpy as np
import matplotlib.pyplot as plt

def s(t):
    return 4.9 * t**2

for h in [1, 0.5, 0.1, 0.01, 0.001, 0.0001]:
    v_media = (s(1 + h) - s(1)) / h
    print(f"h = {h:7.4f} s    velocidad media entre t = 1 y t = {1 + h:.4f}:  {v_media:.5f} m/s")
```

Los valores se acercan a 9,8 m/s sin llegar a alcanzarlo para ningún $h$ distinto de cero. Esa cifra, a la que se acerca la velocidad media cuando el intervalo tiende a cero, es la **velocidad instantánea**, y se llama **derivada** de $s$ respecto a $t$: $v = s'(t) = ds/dt$.

## 2. La derivada numérica

Escribimos una función que calcula la derivada de cualquier función con un $h$ pequeño. La versión **centrada** usa la diferencia simétrica $\dfrac{f(x+h) - f(x-h)}{2h}$ y es más precisa que la hacia delante.

```python
def derivada(f, x, h=1e-5):
    """Derivada numérica centrada de f en x."""
    return (f(x + h) - f(x - h)) / (2 * h)

def derivada_adelante(f, x, h=1e-5):
    return (f(x + h) - f(x)) / h

print("derivada centrada:  ", derivada(s, 1.0))
print("derivada adelante:  ", derivada_adelante(s, 1.0))
print("valor exacto 9,8 t: ", 9.8 * 1.0)
```

## 3. Cómo depende el error del paso

Si $h$ es demasiado grande, la aproximación es mala; si es demasiado pequeño, el ordenador resta números casi iguales y pierde precisión. Vamos a verlo con una función cuya derivada conocemos, $f(x) = x^3$, con derivada $3x^2$.

```python
f = lambda x: x**3
x0 = 2.0
exacta = 3 * x0**2
hs = np.logspace(-12, 0, 60)
err_ad = np.abs([derivada_adelante(f, x0, h) - exacta for h in hs])
err_ce = np.abs([derivada(f, x0, h) - exacta for h in hs])

plt.figure(figsize=(7, 4))
plt.loglog(hs, err_ad, label="hacia delante")
plt.loglog(hs, err_ce, label="centrada")
plt.xlabel("paso h"); plt.ylabel("error absoluto"); plt.legend(); plt.grid(alpha=0.3, which="both")
plt.title("Error de la derivada numérica de x³ en x = 2")
plt.show()
```

Verás una recta descendente (el error baja con $h$) hasta que alrededor de $h \approx 10^{-6}$ a $10^{-8}$ sube de nuevo: es el límite de la precisión del ordenador. Un paso de $10^{-5}$ es una buena elección.

## 4. Aplicación al cielo: ¿a qué ritmo se mueve el Sol hacia el norte y el sur?

La declinación del Sol que usamos en la semana 1 es una función del día $d$: $\delta(d) = -23{,}44^\circ\cos\!\big(360^\circ (d+10)/365\big)$. Su derivada es el ritmo diario de cambio.

```python
def declinacion(d):
    return -23.44 * np.cos(np.radians(360 / 365 * (d + 10)))

dias = np.arange(1, 366)
ritmo = np.array([derivada(declinacion, d, 1e-3) for d in dias])      # °/día

fig, axs = plt.subplots(2, 1, figsize=(9, 5), sharex=True)
axs[0].plot(dias, declinacion(dias)); axs[0].set_ylabel("declinación (°)"); axs[0].grid(alpha=0.3)
axs[1].plot(dias, ritmo); axs[1].set_ylabel("ritmo (°/día)"); axs[1].set_xlabel("día del año"); axs[1].grid(alpha=0.3)
axs[1].axhline(0, color="gray", lw=1)
plt.show()

print(f"Ritmo máximo: {ritmo.max():.3f} °/día (día {dias[ritmo.argmax()]})")
print(f"Ritmo en el solsticio de junio (día 172): {derivada(declinacion, 172, 1e-3):+.4f} °/día")
```

La declinación cambia más deprisa en los equinoccios (unos 0,4° al día, cerca de tres cuartas partes del diámetro del Sol cada día) y casi se detiene en los solsticios: donde la derivada vale cero la función tiene un máximo o un mínimo. La palabra *solsticio* significa literalmente «Sol quieto».
