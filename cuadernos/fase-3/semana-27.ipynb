## 1. Una onda en el espacio y en el tiempo

$$y(x, t) = A\sin(kx - \omega t), \qquad k = \frac{2\pi}{\lambda}, \qquad \omega = \frac{2\pi}{T}.$$

A la izquierda, varias **fotos** de la onda en instantes sucesivos: el patrón se desplaza hacia la derecha. A la derecha, el **registro** de un punto fijo a lo largo del tiempo: sube y baja en el mismo sitio.

```python
import numpy as np
import matplotlib.pyplot as plt

A, lam, T = 1.0, 2.0, 1.0            # amplitud, longitud de onda (m), periodo (s)
k, w = 2 * np.pi / lam, 2 * np.pi / T
print(f"velocidad de la onda: v = λ/T = {lam / T} m/s")

x = np.linspace(0, 6, 400)
t = np.linspace(0, 3, 400)
fig, (a1, a2) = plt.subplots(1, 2, figsize=(12, 3.8))
for ti in [0, 0.1, 0.2, 0.3]:
    a1.plot(x, A * np.sin(k * x - w * ti), label=f"t = {ti} s")
a1.set_xlabel("x (m)"); a1.set_title("Fotos en el espacio"); a1.legend(fontsize=8); a1.grid(alpha=0.3)
a2.plot(t, A * np.sin(k * 1.0 - w * t))
a2.set_xlabel("t (s)"); a2.set_title("Registro en el punto x = 1 m"); a2.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

## 2. Superposición: sumar ondas

Tres casos con dos ondas de amplitud 1: en fase, en oposición y con un desfase de 90°.

```python
t = np.linspace(0, 2, 500)
w = 2 * np.pi
fig, axs = plt.subplots(1, 3, figsize=(13, 3.2), sharey=True)
for ax, (desfase, nombre) in zip(axs, [(0, "en fase"), (np.pi, "en oposición"), (np.pi / 2, "desfase de 90°")]):
    y1, y2 = np.sin(w * t), np.sin(w * t + desfase)
    ax.plot(t, y1, lw=0.8, label="onda 1"); ax.plot(t, y2, lw=0.8, label="onda 2")
    ax.plot(t, y1 + y2, "k", lw=2, label="suma")
    ax.set_title(f"{nombre}: amplitud {np.abs(y1 + y2).max():.2f}"); ax.grid(alpha=0.3)
axs[0].legend(fontsize=8); plt.tight_layout(); plt.show()
```

Con un desfase de 90°, la amplitud sale $\sqrt{2} \approx 1{,}41$: las dos ondas se suman como dos vectores perpendiculares (semana 9).

## 3. Pulsaciones

Dos frecuencias parecidas, 10 y 11 Hz. La suma oscila deprisa, a unos 10,5 Hz, con una amplitud que sube y baja una vez por segundo.

```python
t = np.linspace(0, 3, 6000)
f1, f2 = 10, 11
suma = np.sin(2 * np.pi * f1 * t) + np.sin(2 * np.pi * f2 * t)
envolvente = 2 * np.abs(np.cos(np.pi * (f1 - f2) * t))
plt.figure(figsize=(11, 3.5))
plt.plot(t, suma, lw=0.7)
plt.plot(t, envolvente, "r--", lw=1, label="envolvente"); plt.plot(t, -envolvente, "r--", lw=1)
plt.xlabel("t (s)"); plt.title(f"Pulsaciones: {abs(f1 - f2)} por segundo"); plt.legend(); plt.grid(alpha=0.3)
plt.show()
```

## 4. Las franjas de Young

Dos rendijas separadas una distancia $d$, y una pantalla a una distancia $L$. A un punto de la pantalla a una altura $y$, las distancias desde cada rendija son

$$r_{1,2} = \sqrt{L^2 + (y \mp d/2)^2}$$

(Pitágoras, semana 7). Las dos ondas llegan con un desfase $k\,(r_2 - r_1)$, y la intensidad (que va con el cuadrado de la amplitud) es proporcional a $\cos^2\left(\tfrac{k\,(r_2 - r_1)}{2}\right)$. Lo calculamos exactamente y lo comparamos con la predicción $\Delta y = \lambda L/d$.

¿De dónde sale esa predicción? Si $L$ es mucho mayor que $d$ y que $y$, la diferencia de caminos es aproximadamente $d\,y/L$ (es el cateto pequeño de un triángulo semejante al que forman $y$ y $L$). Hay franja brillante cuando esa diferencia es un número entero de longitudes de onda: $d\,y/L = m\lambda$, es decir, $y = m\lambda L/d$.

```python
lam = 530e-9          # luz verde, m
d = 0.1e-3            # separación entre rendijas, m
L = 1.0               # distancia a la pantalla, m
k = 2 * np.pi / lam

y = np.linspace(-0.02, 0.02, 4000)     # posiciones en la pantalla, m
r1 = np.sqrt(L**2 + (y - d / 2) ** 2)
r2 = np.sqrt(L**2 + (y + d / 2) ** 2)
intensidad = np.cos(k * (r2 - r1) / 2) ** 2

plt.figure(figsize=(11, 3.5))
plt.plot(y * 1000, intensidad)
for m in range(-3, 4):
    plt.axvline(m * lam * L / d * 1000, color="r", lw=0.6, ls="--")
plt.xlabel("posición en la pantalla (mm)"); plt.ylabel("intensidad relativa")
plt.title(f"Franjas de Young: separación prevista λL/d = {lam * L / d * 1000:.2f} mm"); plt.show()
```

Las líneas rojas (la predicción aproximada) caen sobre los máximos calculados con la fórmula exacta. En un experimento real, además, cada rendija tiene una anchura que difracta la luz, y la intensidad de las franjas va disminuyendo hacia los lados.

## 5. Dos fuentes en un estanque

Por último, el patrón completo en un plano: dos fuentes puntuales que emiten en fase, como dos piedras que caen a la vez. En cada punto sumamos las dos ondas, con amplitud que disminuye con la distancia. Las zonas de interferencia destructiva forman líneas que salen en abanico.

```python
lam = 1.0
k = 2 * np.pi / lam
X, Y = np.meshgrid(np.linspace(-8, 8, 600), np.linspace(0.05, 12, 450))
fuentes = [(-1.5, 0.0), (1.5, 0.0)]
onda = sum(np.cos(k * np.hypot(X - fx, Y - fy)) / np.sqrt(np.hypot(X - fx, Y - fy)) for fx, fy in fuentes)

plt.figure(figsize=(8, 6))
plt.imshow(onda, extent=(-8, 8, 0.05, 12), origin="lower", cmap="RdBu", vmin=-1, vmax=1)
plt.plot([f[0] for f in fuentes], [f[1] for f in fuentes], "ko")
plt.title("Interferencia de dos fuentes (foto en un instante)"); plt.xlabel("x"); plt.ylabel("y")
plt.show()
```
